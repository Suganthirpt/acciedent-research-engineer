# MorTH DATASET EVALUATION

```python
import pdfplumber
import pandas as pd

pdf_path = "../data/raw/morth/RA_2018.pdf"

truck_tables = []

with pdfplumber.open(pdf_path) as pdf:
    for page_num, page in enumerate(pdf.pages):
        
        tables = page.extract_tables()
        
        for table in tables:

            if table and len(table) > 1:
                header_row = [str(cell).lower() for cell in table[0] if cell]

                if any("vehicle" in cell for cell in header_row):
                    df_table = pd.DataFrame(table[1:], columns=table[0])
                    
                    # Filter Truck + Heavy rows
                    df_filtered = df_table[
                        df_table.iloc[:,0].astype(str).str.contains(
                            "truck|heavy", 
                            case=False, 
                            na=False
                        )
                    ]
                    
                    if not df_filtered.empty:
                        df_filtered["Source_Page"] = page_num + 1
                        truck_tables.append(df_filtered)

df_truck = pd.concat(truck_tables, ignore_index=True)

df_truck
```

# MERGING KAGGLE DATASETS

```python

# CONFIGURATION

OUTPUT_FILE = '../data/intermediate/Kaggle_Accidents.csv'

GOLDEN_COLUMNS = [
    'accident_id', 'source', 
    'year', 'month', 'time', 'day_of_week', 
    'state', 
    'party_age_group',        
    'victim_age_group',       
    'vehicle_type', 'vehicles_involved',      
    'cause_of_accident',                      
    'weather_condition', 'road_condition', 'lighting_condition', 
    'casualties', 'severity', 'fatalities'
]


# HELPER: TIME CLEANER

def clean_time_format(val):
    val = str(val).strip()
    if val.lower() in ['nan', 'unknown', '', 'na']: return np.nan
    try:
        dt = pd.to_datetime(val, format='%H:%M:%S', errors='coerce')
        if pd.isnull(dt): dt = pd.to_datetime(val, format='%H:%M', errors='coerce')
        return dt.strftime('%H:%M') if pd.notnull(dt) else np.nan
    except: return np.nan


# HELPER: AGE BINNING

def bin_age(value):
    str_val = str(value).lower().strip()
    if str_val in ['nan', 'unknown', '', 'na']: return 'Unknown'
    if '-' in str_val or 'over' in str_val or 'under' in str_val: return str_val.title()
    try:
        age = float(str_val)
        if age < 18: return 'Under 18'
        elif 18 <= age <= 30: return '18-30'
        elif 31 <= age <= 50: return '31-50'
        elif age > 50: return 'Over 51'
        else: return 'Unknown'
    except ValueError: return 'Unknown'


# TRANSFORM DATASET 1

print("Mapping Dataset 1...")
df1_clean = df_kaggle_truck_1.copy()

df1_clean = df1_clean.rename(columns={
    'State Name': 'state',
    'Year': 'year',
    'Month': 'month',
    'Time of Day': 'time',
    'Day of Week': 'day_of_week',
    'Driver Age': 'party_age_raw',
    'Vehicle Type Involved': 'vehicle_type',
    'Weather Conditions': 'weather_condition',
    'Road Condition': 'road_condition',
    'Lighting Conditions': 'lighting_condition',
    'Number of Vehicles Involved': 'vehicles_involved',
    'Number of Casualties': 'casualties',
    'Accident Severity': 'severity',
    'Number of Fatalities': 'fatalities'
})

df1_clean['cause_of_accident'] = np.nan
df1_clean['victim_age_raw'] = np.nan
df1_clean['source'] = 'Kaggle_Accident_prediction'

# ==============================
# TRANSFORM DATASET 2
# ==============================
print("Mapping Dataset 2...")
df2_clean = df_kaggle_truck_2.copy()

df2_clean = df2_clean.rename(columns={
    'Cause_of_accident': 'cause_of_accident',
    'Age_band_of_casualty': 'victim_age_raw',
    'Time': 'time',
    'Day_of_week': 'day_of_week',
    'Age_band_of_driver': 'party_age_raw',
    'Type_of_vehicle': 'vehicle_type',
    'Weather_conditions': 'weather_condition',
    'Road_surface_conditions': 'road_condition',
    'Light_conditions': 'lighting_condition',
    'Number_of_vehicles_involved': 'vehicles_involved',
    'Number_of_casualties': 'casualties',
    'Accident_severity': 'severity'
})

df2_clean['year'] = np.nan
df2_clean['month'] = np.nan
df2_clean['state'] = np.nan
df2_clean['fatalities'] = np.nan
df2_clean['source'] = 'Kaggle_road_csv'

# ==============================
# MERGE & FINALIZE
# ==============================
print("Merging and Normalizing...")
full_df = pd.concat([df1_clean, df2_clean], ignore_index=True)

# 1. Clean Time
full_df['time'] = full_df['time'].apply(clean_time_format)

# 2. Bin Ages
full_df['party_age_group'] = full_df['party_age_raw'].apply(bin_age)
full_df['victim_age_group'] = full_df['victim_age_raw'].apply(bin_age)

# 3. STANDARDIZE SEVERITY (The Specific Mapping)
# We map both numeric (Set 1) and string (Set 2) values to your desired output
severity_map = {
    # Dataset 1 (Numeric)
    1: 'Fatal', '1': 'Fatal',
    2: 'Serious', '2': 'Serious',
    3: 'Minor', '3': 'Minor',       
    
    # Dataset 2 (String)
    'Fatal injury': 'Fatal',
    'Serious Injury': 'Serious',
    'Slight Injury': 'Minor',       
    'Fatal': 'Fatal',              
    'Serious': 'Serious',          
    'Minor': 'Minor'               
}

# Apply map and lowercase cleanup first to match strings reliably
full_df['severity'] = full_df['severity'].replace(severity_map)

# 4. Generate ID
full_df['accident_id'] = [f"KAG_{i:06d}" for i in range(1, len(full_df)+1)]

# 5. Fill Unknowns & Fix Case
for col in GOLDEN_COLUMNS:
    if col not in full_df.columns: full_df[col] = np.nan

text_columns = ['state', 'weather_condition', 'road_condition', 
                'lighting_condition', 'cause_of_accident', 'vehicle_type']

for col in text_columns:
    full_df[col] = full_df[col].fillna('Unknown').replace({'nan': 'Unknown', '': 'Unknown'})
    full_df[col] = full_df[col].astype(str).str.title()

# Save
final_df = full_df[GOLDEN_COLUMNS]
final_df.to_csv(OUTPUT_FILE, index=False)

print(f"SUCCESS! Created '{OUTPUT_FILE}' with {len(final_df)} rows.")
print(final_df['severity'].value_counts()) # Check the new categories
```

```python

# ==============================
# 1. CONFIGURATION
# ==============================
OUTPUT_FILE = '../data/intermediate/social_media_reported_accidents.csv'

# FINAL GOLDEN SCHEMA (Added Year/Month)
GOLDEN_COLUMNS = [
    'accident_id', 'source', 
    'year', 'month', 'date', 'day_of_week', 
    'state', 'city_area', 
    'fatalities', 'injuries', 'casualties', 
    'vehicle_1', 'vehicle_2', 
    'victim_age_group',       
    'cause_of_accident',      
    'weather_condition', 
    'url'
]

# ==============================
# HELPER FUNCTIONS
# ==============================
def bin_age(value):
    str_val = str(value).lower().strip()
    if str_val in ['nan', 'unknown', '', 'na']: return 'Unknown'
    if '-' in str_val or 'over' in str_val or 'under' in str_val: return str_val.title()
    try:
        age = float(str_val)
        if age < 18: return 'Under 18'
        elif 18 <= age <= 30: return '18-30'
        elif 31 <= age <= 50: return '31-50'
        elif age > 50: return 'Over 51'
        else: return 'Unknown'
    except ValueError: return 'Unknown'

def clean_time_format(val):
    val = str(val).strip()
    if val.lower() in ['nan', 'unknown', '', 'na']: return np.nan
    try:
        dt = pd.to_datetime(val, format='%H:%M:%S', errors='coerce')
        if pd.isnull(dt): dt = pd.to_datetime(val, format='%H:%M', errors='coerce')
        return dt.strftime('%H:%M') if pd.notnull(dt) else np.nan
    except: return np.nan

# ==============================
# 2. PROCESS NEWSPAPER DATA
# ==============================
print("Processing Newspaper Data...")
df_paper_clean = merged_df.copy()

df_paper_clean = df_paper_clean.rename(columns={
    'Crash Date': 'date',
    'Crash Day': 'day_of_week',
    'State': 'state',
    'Location': 'city_area',
    'Killed': 'fatalities',
    'Injured': 'injuries',
    'Age': 'victim_age_raw',         
    'Crash Type': 'cause_of_accident', 
    'Vehicle 1': 'vehicle_1',
    'Vehicle/Object 2': 'vehicle_2'
})
df_paper_clean['source'] = 'News_Paper'

# ==============================
# 3. PROCESS NEWS API DATA
# ==============================
print("Processing News API Data...")
df_api_clean = df_news_api.copy()

df_api_clean = df_api_clean.rename(columns={
    'Date': 'date',
    'Title': 'cause_of_accident',      
    'Killed': 'fatalities',
    'Injured': 'injuries',
    'Location': 'city_area',
    'URL': 'url',
    'Vehicle_1': 'vehicle_1',
    'Vehicle_2': 'vehicle_2'
})
df_api_clean['source'] = 'News_API'

# ==============================
# 4. PROCESS YOUTUBE DATA
# ==============================
print("Processing YouTube Data...")
df_yt_clean = df_youtube.copy()

df_yt_clean = df_yt_clean.rename(columns={
    'Date': 'date',
    'Title': 'cause_of_accident',      
    'Killed': 'fatalities',
    'Injured': 'injuries',
    'State': 'state',
    'District ': 'city_area',
    'URL': 'url',
    'Vehicle 1': 'vehicle_1',
    'Vehicle 2 ': 'vehicle_2' 
})
df_yt_clean['source'] = 'YouTube'

# ==============================
# 5. MERGE & STANDARDIZE
# ==============================
print("Merging Datasets...")
full_df = pd.concat([df_paper_clean, df_api_clean, df_yt_clean,df_new_clean], ignore_index=True)

# --- Standardize Columns ---

# 1. Fix Dates (YYYY-MM-DD) & Derive Year/Month
full_df['date'] = pd.to_datetime(full_df['date'], errors='coerce')
full_df['year'] = full_df['date'].dt.year
full_df['month'] = full_df['date'].dt.month_name() # e.g., 'January'
full_df['date'] = full_df['date'].dt.strftime('%Y-%m-%d') # Stringify for CSV

# 2. Calculate Day of Week (if missing)
full_df['day_of_week'] = full_df['day_of_week'].fillna(
    pd.to_datetime(full_df['date']).dt.day_name()
)

# 3. Clean Numbers
for col in ['fatalities', 'injuries']:
    full_df[col] = pd.to_numeric(full_df[col], errors='coerce').fillna(0).astype(int)

# 4. Calculate Casualties
full_df['casualties'] = full_df['fatalities'] + full_df['injuries']

# 5. Bin Ages
if 'victim_age_raw' not in full_df.columns:
    full_df['victim_age_raw'] = np.nan
full_df['victim_age_group'] = full_df['victim_age_raw'].apply(bin_age)

# 6. Generate IDs
full_df['accident_id'] = [f"SCRAPED_{i:06d}" for i in range(1, len(full_df)+1)]

# 7. Ensure all Golden Columns exist
for col in GOLDEN_COLUMNS:
    if col not in full_df.columns:
        full_df[col] = np.nan

# 8. Text Cleanup (Fill Unknowns)
text_cols = ['state', 'city_area', 'cause_of_accident', 'weather_condition', 'vehicle_1', 'vehicle_2']
for col in text_cols:
    full_df[col] = full_df[col].fillna('Unknown').replace({'nan': 'Unknown', '': 'Unknown'})
    full_df[col] = full_df[col].astype(str).str.title().str.strip()

# ==============================
# 6. SAVE
# ==============================
final_df = full_df[GOLDEN_COLUMNS]
final_df.to_csv(OUTPUT_FILE, index=False)

print(f"final_df_SUCCESS! Merged {len(final_df)} records.")
```
